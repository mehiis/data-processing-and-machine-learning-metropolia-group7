<b>Satunnaismetsät</b> jatkavat <b>päätöspuiden</b> logiikkaa ja toiminnallisuutta eteenpäin. Viime oppitunnilla kävimme läpi <b>päätöspuut</b> (Decision trees), jotka ovat puualgoritmin (tree algorithm) muotoisia luokittelijoita (classifiers), jossa jokainen solmu (node) sisältää datasetin ominaisuuden, ja oksat (branch) määrittelevät kuinka puussa liikutaan solmusta toiseen.

Tässä kappaleessa käymme läpi mikä on päätöspuiden heikkous ja kuinka <b>satunnaismetsät</b> (random forests) auttavat korjaamaan tätä ongelmaa. Ongelmaksi muodostuu kun haarautumisvaihtoehdot ovat hyvin lähellä tosiaan ja valinta tehdään hyvin marginaalisesti, joka  voi johtaa heittelevään dataan joka voi <b>tuntua</b> satunnaiselta. Tällaisissa tilanteissa satunnaismetsät tulevat hyötykäytöön.

<b>Satunnaismetsät ovat kokoelma päätöspuita.</b> Eli sen sijaan, että luomme yhden puun niin niitä generoidaankin useampi (esim. 100 kappaletta) kerralla. Kun päätöspuu kokoelma on valmis, me annamme kokoelman äänestää luokittelun lopputuleman jokaiselle havainnolle. Tällaisia metodeja missä kokoamme useamma mallin kutsutaan <b>yhdistelmämenetelmät</b>. (Ensemble methods)
<center>
<figure>
    
    <figcaption>Kuva 1. Havainnollistava kuva luokittelusta. Kuusi päätöspuuta, joiden data on valittu uusio-otantamenetelmää käyttäen.</figcaption>
</figure>
</center>
Jokaisen päätöspuun lajittelun lopputulos on ennalta päätetty (deterministinen), joten kaikkien metsän puiden lopputulos olisi sama jos emme koodilla loisi siihen satunnaisuutta. <b> Näitä kahta tapaa yhdistäen luodaan satunnaisuutta päätöspuiden lopputulokseen:</b>

1. Ensin teemme alkuperäisestä tietojoukosta uuden saman kokoisen opetusdatan poimimalla sieltä satunnaisesti datapisteitä palauttaen, jolloin osa datapisteistä toistuu uudessa datassa ja osa jää kokonaan pois. (Vaihe 1)
2. Ja lopuksi valitsemme <b>satunnaisesti ominaisuuksia</b>, joiden avulla rakennamme puun. (Vaihe 2)

<br/>
<b>Satunnasmetsän pseudoalgoritmi:</b>
</br>
<i><b>n</b> = Päätöspuiden määrää satunnaismetsässä. </i><br/>

<code>
#koulutusvaihe
For i = 1 to n {
  1. Luo uusi päätöspuu käyttäen uusi-otantaa. (bootstrap sampling)
  2. Satunnaisesti valitse osajoukko ominaisuuksia aina kun puu haarautuu.
  3. Rakenna päätöspuu i.
}

#ennustusvaihe
For each havainto (observation) {
    For i = 1 to n {
        Luokittele havainto (observation) puulla i ja tallenna sen ääni (vote_i).
    }
    Aseta lopulliseksi tulokseksi (outcome) eniten ääniä saanut luokka tai ennusteen keskiarvo.
}
</code>

<br/>

<b>Uusio-otanta (boostrap sampling)</b> <br/>
Uusio-otanto toimii niin, että jos meillä on esimerkiksi .csv data jossa on 100 kpl rivejä (datapisteitä) valitsemme sieltä 100 kpl rivejä niin, että palautamme sen takaisin tietopooliin (datapool). Tällä tavalla päätöspuu on rakennettu niin, että siellä on samoja toistuvia arvoja. <b>esim. rivi 6 on päätöspuussa 10 kertaa.</b>

---

<h2>Satunnaismetsä esimerkki Iris dataa käyttäen</h2>
Rakennamme ensimmäisen satunnaispuun käytäen jo meille tuttua Iris dataa.

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
import pandas as pd

# ladataan iris-tietoaineisto
iris = load_iris()
X = iris.data
y = iris.target

# luodaan satunnaismetsä luokittelija olio
satunnaismetsa = RandomForestClassifier(n_estimators=100, bootstrap=True, max_samples=0.7, max_features=0.75, random_state=123)

# Carry out 10-fold cross-validation
cv_scores = cross_val_score(satunnaismetsa, X, y, cv=10)

# näytä tulokset
print(f'Fold-specific accuracies: {cv_scores}')
print(f'Overall accuracy (avg. of folds): {cv_scores.mean():.4f}')
```

<br/>
<b>sklearn</b> kirjastossa satunnasipuut rakennetaan käyttäen <i>RandomForestClassifier</i> luokkaa.
<br/>

---

<u><b>Satunnaispuun parametrit:<b></u>

<b>n_estimators (int):</b> Määrittää päätöspuiden määrää satunnaismetsässä. Esimerkissä meillä on 100 päätöspuuta. (Oletus 100 puuta, jos jättää tyhjäksi.)

- Paljon puita johtaa vakaampaan ja tarkempaan tulokseen, mutta vaatii enemmän laskentatehoa.

<b>bootstrap (boolean):</b> Käytetäänkö uusio-otantamenetelmää (true/false). Uusio-otantamenetelmä ehkäisee ylioppimista ja tekee puista mahdollisimman monimuotoisia. (Oletus on true)

<b>max_samples (float):</b> <i>(Vain jos bootstrap on 'true')</i> Säätää, kuinka suuri suhteellinen osuus (esim. 0.70 = 70 %) tai absoluuttinen määrä opetusdatan riveistä poimitaan kullekin puulle silloin. Pienempi arvo nopeuttaa mallin opetusta.

<b>max_features (float):</b> Määrittää kuinka monta satunnaista ominaisuutta otetaan tarkasteluun haarautumisvaiheessa (Vaihe 2). Luku 0.75 tarkoittaa, että huomiomme 75% kaikista ominaisuuksista eli $\frac{1}{4}$ ominaisuuksista jää pois. 
- Pienempi arvo pienentää haarojen korrelaatiota.

<b>random_state (int):</b> Siemenluku satunnaisuudelle; tämä luo satunnaisluvun joka on deterministinen, eli voimme toistaa satunnaisuuden ja lopputulos on sama.

---

<i>Lisäksi on vielä mahdollista muokata päätöspuiden rakennetta. Esimerkiksi:</i><br/>
<b>max_depth parametri:</b> Hallitsee puiden maksimisyvyyttä. Oletusarvolla (None) puut kasvavat niin syviksi, että lehtisolmut ovat täysin puhtaita tai niissä on alle tietyn vähimmäismäärän (min_samples_split) havaintoja.

---

<b>The cross_val_score:</b> Suorittaa 10-osaisen ristikkäisvalidoinnin (10-fold cross-validation). Data jaetaan 10 yhtä suureen osaan (Folds). Malli opetetaan 9 osalla ja testataan 1 osalla, ja tämä toistetaan 10 kertaa niin, että jokainen osa toimii vuorollaan testidatana (Test data). Funktion palauttamien tarkkuuksien (Accuracy) keskiarvo (Average) antaa realistisen kuvan mallin yleisestä suorituskyvystä.

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

model.fit(X, y)

# Number of trees to plot
n_trees_to_plot = 5

# Create a figure to hold the plots
plt.figure(figsize=(20, 10))

for i in range(n_trees_to_plot):
    plt.subplot(1, n_trees_to_plot, i + 1)  # 1 row and n_trees_to_plot columns
    plot_tree(
            model.estimators_[i], 
            filled=True,
            feature_names=iris.feature_names, 
            class_names=list(iris.target_names)
             )
    plt.title(f'Tree {i + 1}')

plt.tight_layout()
plt.show()
```

Tältä stunnaismetsämalli näyttää. On mahdollista visualisoida (Visualize) metsän päätöspuut (Decision trees). Ajetaan analyysi uudelleen ja tarkastellaan metsän viittä ensimmäistä puuta. Yksinkertaisuuden vuoksi tässä esimerkissä ei suoriteta ristikkäisvalidointia (Cross-validation).

<h3>Datapisteiden ominaisuuksien tärkeys</h3>

Satunnaismetsät voivat myös mitata ominaisuuksien tärkeyttä. <b>Ominaisuuden tärkeys kertoo, kuinka paljon kukin ominaisuus vaikuttaa mallin ennusteisiin.</b> Ne lasketaan 'Decrease in impurity' -metodilla (esim. Gini impurity), jonka kukin ominaisuus lasketaan solmujen jaossa. Mitä enemmän ominaisuus vähentää epäpuhtautta, sitä tärkeämpänä sitä pidetään (Eli mitä lähempänä ykköstä).

<b>Ominaisuuksien tärkeysarvot normalisoidaan niin, että niiden summa on 1. Mitä korkeampi arvo on, sitä tärkeämpi ominaisuus on.</b> Lasketaan Iris-tietojoukon ominaisuuksien tärkeys:

```python
print (model.feature_importances_)
```

<b>0.01426822 + 0.01436214 + 0.51040072 + 0.46096892 = 1</b>

---

<b>Tulos osoittaa, että ominaisuudet kolme ja neljä ovat ylivoimaisesti tärkeimpiä satunnaismetsämallin ennusteille.</b> 
- Kolmas ominaisuus vaikuttaa noin 51 %.
- Neljäs ominaisuus vaikuttaa noin 46 %.

Tämä indikoi, että näillä kahdella ominaisuudella (3 ja 4) on huomattava vaikutus mallin päätöksiin. Ensimmäisellä ja toisella ominaisuudella on paljon alhaisemmat tärkeyspisteet (noin 1,4 % kukin), eli niiden panos ennusteen lopputuloksessa on vähäinen.

<h3>Satunnaismetsät ja validaatio</h3>


```python
# read data (replace with your own path)
df = pd.read_csv('../../datasets/titanic_v2.csv', sep=';')
# drop rows with missing values
df.dropna(axis=0, how='any', inplace=True)
display(df)
```

```python
colnames = df.columns
colnames
```

```python
# type conversions
# Note that DecisionTreeClassifier can't use strings in explanatory variables
df['sex'] = df['sex'].map({'male': 1, 'female': 2})
df.describe()
```

```python
# plot histograms of the variables
df.hist(figsize=(10, 10))
plt.show()
```

```python

# extract explanatory variables into a data frame
X = df.loc[:, 'pclass':'age']
# extract response variable (class variable) into a series
Y = df.loc[:, 'survived']

# Split the data into training and test sets
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=123)
```

```python
# Create a random forest classifier and train it
model = RandomForestClassifier(n_estimators=100, bootstrap=True, max_samples=0.7, min_samples_leaf=30, random_state=123)
model.fit(X_train, Y_train)
```

```python
# Display the confusion matrix and the classification report
from sklearn.metrics import accuracy_score, confusion_matrix
Y_pred = model.predict(X_test)
print(f'Confusion matrix:\n{confusion_matrix(Y_test, Y_pred)}')
from sklearn.metrics import classification_report
print(classification_report(Y_test, Y_pred))

# Display the feature importances
print(model.feature_importances_)
```

```python
# Display the feature importances
print(model.feature_importances_)

# Peek a few trees
plt.figure(figsize=(20, 10))
for i in range(5):
    plt.subplot(1, 5, i + 1)
    plot_tree(
        model.estimators_[i], 
        filled=True, 
        feature_names=list(colnames[:-1]), 
        class_names=['0', '1']
    )
    plt.title(f'Tree {i + 1}')
plt.tight_layout()
plt.show()

# Output the first tree into a file
from sklearn.tree import export_text
tree_text = export_text(model.estimators_[0], feature_names=colnames[:-1].tolist())
with open('tree.txt', 'w') as f:
    f.write(tree_text)
```
