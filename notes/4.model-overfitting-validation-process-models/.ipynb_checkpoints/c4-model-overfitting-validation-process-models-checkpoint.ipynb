<h3>Ylisovittaminen (Model overfitting)</h3>

<i>"Ylisovittaminen (overfitting) tarkoittaa tilannetta, jossa malli oppii harjoitusdatan kohinan ja yksityiskohdat varsinaisten taustalla olevien mallien sijaan – ja suoriutuu siksi heikosti uudella, aiemmin näkemättömällä datalla."</i> -[Lenovo](https://www.lenovo.com/fi/fi/knowledgebase/regularization-a-comprehensive-guide-to-enhancing-machine-learning-models/)

- Ylisovittaminen tapahtuu, kun malli oppii opetusdatan erityispiirteet tai kohinan (noise) liian tarkasti, jolloin se ei enää löydä yleistettäviä säännönmukaisuuksia (generalizable patterns).
    - Esim. malli voisi oppia ennustamaan vasenkätisyyttä täydellisesti pituuden (height), painon (weight) ja kasvojen pisamien (freckles) perusteella opetusdatassa, mutta epäonnistuisi täysin uuden ihmisryhmän kohdalla.
-  <b>Tämän estämiseksi mallin toimintaa on aina testattava uuden datan avulla, mitä kutsutaan validoinniksi (validation).</b>

---
<h3>Split validation</h3>
Yksinkertaisin ja yleisin tapa on jaettu validointi (split validation), jossa data jaetaan tyypillisesti siten, että <b>kaksi kolmasosaa käytetään opettamiseen (training) ja yksi kolmasosa testaamiseen (testing)</b>.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

iris = pd.read_csv('../../datasets/iris.csv')
```

```python
#Luodaan muuttuja X, johon tallennetaan data-aineiston kaikki sarakkeet paitsi 'species'. 
#Tämä muuttuja edustaa piirteitä (features) eli syötedataa, joiden perusteella koneoppimismalli tekee tulevaisuudessa ennustuksia.
X = iris.drop(columns = 'species')

#Muuttuja y, johon eristetään pelkästään data-aineiston 'species'-sarake. 
#Tämä on kohdemuuttuja eli nimike (label), jota koneoppimismallin on tarkoitus oppia tunnistamaan ja ennustamaan.
y = iris['species']


#train_test_split-funktio: jakaa datan satunnaisesti kahteen osaan: opetusjoukkoon (training set) ja testijoukkoon (test set).

#X_train (Opetusdatan piirteet / Training set features): Sisältää satunnaisesti valitut 70 % alkuperäisestä syötedatasta. 
#Tämä on oppimateriaali, jonka malli saa nähdä koulutusvaiheessa. 
#Se sisältää vain havainnot, joiden pohjalta algoritmi yrittää löytää säännönmukaisuuksia (patterns).

#X_test (Testidatan piirteet): Sisältää loput 30 % syötedatasta. Tämä on mallin "koepaperi". 
#Testivaiheessa opetettu malli saa nämä täysin uudet tiedot nähtäväkseen, ja sen täytyy soveltaa oppimaansa tehdäkseen ennustuksia niiden pohjalta.

#y_train (Training set target variables): Sisältää 70 % alkuperäisistä oikeista vastauksista eli nimikkeistä (labels). 
#Tämä data on paritettu täsmälleen X_train-muuttujan rivien kanssa. 
#Koulutuksen aikana malli katsoo X_train-piirteitä, arvaa tuloksen ja tarkistaa välittömästi y_train-muuttujasta oliko se oikeassa, säätäen sisäistä logiikkaansa sen mukaan.

#y_test (Test set target variables): Sisältää loput 30 % oikeista vastauksista. 
#Nämä vastaavat X_test-muuttujan rivejä. 
#Tämä on "vastausavain", joka pidetään mallilta ehdottomasti piilossa. 
#Kun malli on antanut ennustuksensa X_test-datan perusteella, tehtyjä arvauksia verrataan tähän y_test-muuttujaan onnistumisprosentin ja mallin todellisen yleistyskyvyn (generalization capability) laskemiseksi.

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=123)

#X, y: Annetaan funktiolle käsiteltäväksi aiemmin määritellyt piirteet ja kohdemuuttuja.

#test_size=0.3: Parametri määrittää, että 30% datasta varataan testijoukoksi mallin arviointia varten. 
#Loput 70 % datasta jäävät opetusjoukoksi.

#random_state=123: Satunnaislukugeneraattorin siemenluvun asettaminen varmistaa, että datan jako on toistettava. 
#Koodi tuottaa täsmälleen saman, satunnaistetun jaon jokaisella suorituskerralla.
```

Satunnaisuudesta johtuen tulokset voivat vaihdella hieman. <b>random_state</b>-parametrin kiinnittäminen (esim. <i>123</i>) varmistaa toistettavuuden (reproducibility), mutta arvoja ei pidä manipuloida suotuisten tulosten saamiseksi (cherry picking).
- Opetusdatan (training data) antama tarkkuus (accuracy) on yleensä paljon korkeampi kuin testidatan, joten <b>sitä ei pidä pitää todellisena suorituskyvyn mittarina.</b>
- Jaetun validoinnin (split validation) heikkous on, että <b>osa datasta joudutaan uhraamaan testaukseen, mikä on ongelmallista pienillä tietojoukoilla.</b>
- Ylisovittamisen (model overfitting) riski kasvaa myös sitä mukaa, kun ominaisuuksien (features) ja mallin monimutkaisuuden määrä lisääntyy.

```python
#Luodaan k:n lähimmän naapurin luokittelija (k-Nearest Neighbors classifier) ja tallennetaan se muuttujaan knn. 
#Parametri "n_neighbors=3" määrittää, että algoritmi tarkastelee kolmea lähintä datapistettä (data point) tehdessään päätöstä uuden havainnon luokasta. 
#Tämä on mallin alustusvaihe (initialization), jossa määritellään mallin rakenne ja toimintasäännöt ennen varsinaisen datan antamista
knn = KNeighborsClassifier(n_neighbors=3)

#Tällä rivillä luokittelija (classifier) opetetaan opetusdatan avulla. 
#Metodi fit ottaa vastaan opetusjoukon piirteet (training set features) X_train sekä opetusjoukon oikeat kohdemuuttujat y_train. 
#K-NN -algoritmin (k-NN algorithm) tapauksessa malli (model) käytännössä tallentaa tämän datan (data) muistiinsa (memory), 
#jotta se voi myöhemmin laskea etäisyyksiä (calculate distances) uusien näytteiden (samples) ja näiden tunnettujen pisteiden välillä.
knn.fit(X_train, y_train)

#Tässä koodirivissä opetettu malli (trained model) tekee ennustuksia uudelle, aiemmin näkemättömälle testijoukolle. 
#Metodi "predict" saa syötteekseen testidatan piirteet (test data features) X_test. 
#Malli etsii jokaiselle testijoukon alkiolle kolme lähintä naapuria tallennetusta opetusdatasta ja päättelee niiden enemmistön perusteella alkion luokan. 
#Nämä ennustetut nimikkeet (predicted labels) tallennetaan taulukkona uuteen muuttujaan y_pred.
y_pred = knn.predict(X_test)

#Tällä rivillä lasketaan mallin tekemien ennustusten tarkkuus (accuracy). 
#Funktio accuracy_score vertaa testijoukon todellisia, oikeita kohdemuuttujia (true target variables) y_test mallin juuri antamiin ennustuksiin y_pred. 
#Funktio palauttaa numeerisen arvon 0.0 ja 1.0 väliltä.
#Tämä kuvaa täsmälleen oikein ennustettujen arvojen osuutta (proportion), ja tämä desimaaliluku sijoitetaan muuttujaan accuracy_test.
accuracy_test = accuracy_score(y_test, y_pred)



print(f"Accuracy of kNN classifier on the test set: {accuracy_test:.2f}")
```

<h3>Ristiinvalidointi (Cross-validation)</h3>

Ristiinvalidoinnin (cross-validation) perusajatus on, että kaikkea saatavilla olevaa dataa käytetään vuorotellen sekä mallin opettamiseen  että testaamiseen. <b>Tämä on erityisen hyödyllistä silloin, kun tietojoukko on pieni</b>, sillä perinteinen jaettu validointi (split validation) saattaa tällöin jättää liian vähän materiaalia mallin oppimiseen tai antaa satunnaisen jaon vuoksi vääristyneitä tuloksia.


Ristiinvalidoinnissa opetusdata jaetaan k-määrään yhtä suuria lohkoja (folds). Malli koulutetaan k kertaa siten, että jokaisella kerralla yksi lohko on testidatana (testing) ja loput k-1 lohkoa opetusdatana (training).
- esim. On käytössä 200 näytettä (samples) ja käytetään 10-osaista ristiinvalidoimista (10-fold cross-validation).
  - Data jaetaan 10 lohkoon (20 näytettä kussakin).
  - Jokaisella koulutuskerralla 180 näytettä on opetusdatana ja 20 testidatana.

<br>
<b>Tämän tuloksena jokaista havaintoa (observation) käytetään testaamiseen tasan yhden kerran.</b>

```python
from sklearn.model_selection import cross_val_score

#Funktio suorittaa ristiinvalidoinnin (cross-validation) automaattisesti ja palauttaa mallin suorituskyvyn jokaisesta lohkosta taulukkona. 

#Se ottaa vastaan seuraavat parametrit:
#knn: Arvioitava koneoppimismalli, jota halutaan testata. Tässä tapauksessa kyseessä on aiemmin määritelty k-NN -luokittelija.
#X: Koko tietojoukon syötedata. Funktio hoitaa tämän datan jakamisen lohkoihin itsenäisesti taustalla.
#y: Sisältää koko tietojoukon oikeat kohdemuuttujat eli nimikkeet, joita malli yrittää ennustaa.

#cv=10: Määrittää ristiinvalidoinnin lohkojen lukumäärän. 
#Arvo 10 tarkoittaa, että suoritetaan 10-osainen ristiinvalidointi, jolloin malli opetetaan ja testataan kymmenen erillistä kertaa.

#scoring='accuracy': Määrittää käytettävän suorituskyvyn mittarin (coring metric). 
#Arvo 'accuracy' ohjeistaa funktiota mittaamaan luokittelijan tarkkuutta eli sitä, kuinka suuri osa sen tekemistä ennustuksista on täsmälleen oikein.

#Lopuksi funktio palauttaa numeerisen taulukon (array), joka sisältää kymmenen erillistä tarkkuusarvoa (yksi jokaiselta ristiinvalidoinnin kierrokselta).
scores = cross_val_score(knn, X, y, cv=10, scoring='accuracy')

#laske keskiarvo ja prinnttaa tulos ulos.
average_accuracy = scores.mean()
print(f"Average accuracy of kNN classifier using 10-fold cross-validation: {average_accuracy:.2f}")
```

<h3>Prosessimallit (Process models)</h3>

Prosessimallit ovat jäsenneltyjä viitekehyksiä, jotka auttavat hallitsemaan koneoppimisprojektien monimutkaisuutta. 

Data-analyytikon työ vaatii jatkuvaa kokeilua ja päätöksentekoa siitä, mitä menetelmää käytetään, miten data esikäsitellään (preprocessing), miten malli validoidaan ja miten tulokset arvioidaan.

Ilman selkeää rakennetta tämä prosessi on altis harha-askelille ja umpikujille. 
<br><br>Riskinä on
- Ajan hukkaaminen
- Sekavuus
- Samojen kokeilujen toistaminen samoilla tai eri parametriasetuksilla siten, että tekijä kadottaa käsityksen jo tehdyistä asioista.

<br>Yksityiskohtiin uppoutuessa alkuperäinen tavoite ja kokonaiskuva unohtuvat helposti. Prosessimallit (process models) ratkaisevat tämän ongelman tarjoamalla selkeät askeleet data-analyysiprojektien läpivientiin, jolloin fokus pysyy tiukasti olennaisessa.

<h3>CRISP-DM (Cross-Industry Standard Process for Data Mining)</h3>

<b>Yksi laajimmin käytetyistä prosessimalleista</b> tiedonlouhinnassaja koneoppimisessa. Se tarjoaa selkeän, <b>kuusivaiheisen rakenteen data-analyysiprojektien toteuttamiseen.</b>

Esimerkkejä:
- [CRISP-DM template](https://github.com/MichaelVerdegaal/crisp-dm-template) by: MichaelVerdegaal
- [CRISP-DM Analysis template in Jupyter](https://github.com/patiegm/Datasci_Resources/blob/master/CRISP-DM%20Analysis%20Template.ipynb) by: patiegm

1. <b>Liiketoiminnan ymmärtäminen (Business understanding):</b>

   - Määritellään ratkaistava liiketoimintaongelma (business problem).
   - Tavoitteena on selvittää analyysin päämäärä (goal), vaatimukset (requirements), rajoitteet (constraints) ja odotettu lopputulos (expected outcome).

2. <b>Datan ymmärtäminen (Data understanding):</b> 

   - Kerätään ja tutkitaan saatavilla olevaa dataa.
   - Tässä kartoitetaan datan ominaisuudet, kuten muuttujatyypit ja arvojakaumat, sekä mahdolliset laatuongelmat (quality issues), kuten puuttuvat arvot (missing values), poikkeamat (outliers) ja järjettömät arvot (nonsensical values).

3. <b>Datan esikäsittely (Data preparation):</b>

   - Data siivotaan, muunnetaan ja valitaan olennaiset piirteet (features).
   - Kaikki nämä vaiheet tulee dokumentoida niin tarkasti, että ne ovat myöhemmin täysin toistettavissa.

4. <b>Mallintaminen (Modeling):</b>
    - Valitaan koneoppimismenetelmä, opetetaan malli ja validoidaan se (validation).
    - Dokumentointiin on sisällytettävä tarkasti valittu menetelmä, käytetyt parametrit sekä mitattu mallin suorituskyky.

5. <b>Arviointi (Evaluation):</b>
    - Arvioidaan mallin todellinen onnistuminen.
    - Selvitetään, kuinka hyvin malli suoriutuu ja vastaako se sille alun perin asetettuihin liiketoimintavaatimuksiin (business requirements).<br>

6. <b>Käyttöönotto (Deployment):</b>
    - Malli viedään käytäntöön (deployment).
    - Tähän sisältyy tulosten viestiminen ja suositusten luominen siitä, miten mallia tulisi hyödyntää käytännössä tai mitä jatkotoimenpiteitä tulisi tehdä.

Jupyter-notebook tukee tätä mallia erinomaisesti. Jokainen kuudesta vaiheesta voidaan jakaa omaksi osiokseen muistikirjassa, jolloin koko prosessin dokumentaatio ja koodi pysyvät yhdessä ainoassa tiedostossa. 

Tämä tekee analyysin tulosten jakamisesta ja toistamisesta vaivatonta.
